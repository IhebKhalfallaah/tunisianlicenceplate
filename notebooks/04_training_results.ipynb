```python
import pandas as pd
import matplotlib.pyplot as plt

# Load your CSV (replace 'your_file.csv' with your actual file path)
df = pd.read_csv('/content/drive/MyDrive/Tunis_detection/results.csv')
df.columns = df.columns.str.strip()
print(df.columns)
# Plot Loss curves
plt.figure(figsize=(10, 6))
plt.plot(df['epoch'], df['train/box_loss'], label='Train Box Loss')
plt.plot(df['epoch'], df['val/box_loss'], label='Val Box Loss')
plt.xlabel('Epoch')
plt.ylabel('Loss')
plt.title('Train & Validation Loss per Epoch')
plt.legend()
plt.grid(True)
plt.show()

# Plot mAP (Accuracy-like metric)
plt.figure(figsize=(10, 6))
plt.plot(df['epoch'], df['metrics/mAP_0.5'], label='Accuracy mAP@0.5')
plt.plot(df['epoch'], df['metrics/mAP_0.5:0.95'], label='Accuracy mAP@0.5:0.95')
plt.xlabel('Epoch')
plt.ylabel('mAP')
plt.title('Mean Average Precision per Epoch')
plt.legend()
plt.grid(True)
plt.show()

# Plot Precision & Recall
plt.figure(figsize=(10, 6))
plt.plot(df['epoch'], df['metrics/precision'], label='Precision')
plt.plot(df['epoch'], df['metrics/recall'], label='Recall')
plt.xlabel('Epoch')
plt.ylabel('Value')
plt.title('Precision and Recall per Epoch')
plt.legend()
plt.grid(True)
plt.show()
```
