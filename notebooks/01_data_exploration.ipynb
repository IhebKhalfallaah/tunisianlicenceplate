```python
import os
import random
import shutil

# Set the path to the source folder
source_folder = "/content/train/images"
source_folder_annotation="/content/train/labels"
# Set the path to the destination folder
destination_folder = "/content/custom_data"

# Set the train and validation percentages
train_percentage = 0.8

# Create the train and validation folders
train_folder = os.path.join(destination_folder, "train_imgs")
val_folder = os.path.join(destination_folder, "test_imgs")

images_list = os.listdir(source_folder)
num_train=int(len(images_list)*0.8)

train_images = images_list[:num_train]
val_images = images_list[num_train:]


for image_file in train_images:
      source_path = os.path.join(source_folder, image_file)
      source_path_anno=os.path.join(source_folder_annotation, image_file[:-4]+'.txt')
      destination_path = os.path.join(train_folder, image_file)
      destination_path_anno=os.path.join(train_folder, image_file[:-4]+'.txt')
      shutil.copy(source_path, destination_path)
      shutil.copy(source_path_anno, destination_path_anno)


for image_file in val_images:
      source_path = os.path.join(source_folder, image_file)
      source_path_anno=os.path.join(source_folder_annotation, image_file[:-4]+'.txt')
      destination_path = os.path.join(val_folder, image_file)
      destination_path_anno=os.path.join(val_folder, image_file[:-4]+'.txt')
      shutil.copy(source_path, destination_path)
      shutil.copy(source_path_anno, destination_path_anno)

print("Splitting of images into train, and val folders is complete.")
```

```python
def getimage(index,Train=True):
    folder='/content/custom_data/train_imgs'
    if (not Train):
      folder='/content/custom_data/test_imgs'
    images=os.listdir(folder)
    name=images[index][:-4]
    ann=open(f'{folder}/{name}.txt')
    image=Image.open(f'{folder}/{name}.jpg')
    draw=ImageDraw.Draw(image)
    line=ann.readline()
    while line !="":
        line=line.split(" ")
        x=float(line[1])
        y=float(line[2])
        w=float(line[3])
        h=float(line[4])
        x=int(x*640)
        y=int(y*640)
        w=int(w*640)
        h=int(h*640)
        x=x-w//2
        y=y-h//2
        draw.rectangle((x,y,x+w,y+h),outline="red",width=2)
        line=ann.readline()
    return image
```

```python
start=10
end=20
fig,axes=plt.subplots(1,end-start,figsize=(60,60))
for i in range(start,end):
    axes[i-start].imshow(np.asarray(getimage(i,True)))
```
